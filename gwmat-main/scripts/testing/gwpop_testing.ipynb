```python
import gwpopulation as gwpop
import numpy as np
```

```python
smoothMassDist = gwpop.models.mass.SinglePeakSmoothedMassDistribution(mmin=2.0, mmax= 100.0)

dataset = {
    'mass_1': np.array([36.0, 41.0, 12.0, 34.0]),
    'mass_ratio': np.array([0.8, 0.6, 0.9, 0.95])
}

hyperparameters = {
    'alpha': 3.5,         # Power-law index for m1
    'beta': 1.1,          # Power-law index for q
    'mmin': 5.0,          # Minimum component mass
    'mmax': 85.0,         # Max mass of the power-law part
    'lam': 0.1,           # 10% of events are in the Gaussian peak
    'mpp': 35.0,          # Peak is centered at 35 solar masses
    'sigpp': 4.0,         # Peak has a width of 4 solar masses
    'delta_m': 2.0        # Smooth the distribution over 2 solar masses at the low end
}

probability_density = smoothMassDist(dataset, **hyperparameters)
```

```python
print(probability_density)
```
